## 1. טעינת תמונה ובדיקת מימדים ב-PIL

```python
import os
from PIL import Image

# איתור מיקום התמונה
img_path = "../images/gray_quare_cat.JPG"
if not os.path.exists(img_path):
    img_path = "images/gray_quare_cat.JPG"
if not os.path.exists(img_path):
    import urllib.request
    url = "https://raw.githubusercontent.com/alonitzkovitch/python_course_book/main/images/gray_quare_cat.JPG"
    urllib.request.urlretrieve(url, "cat.jpg")
    img_path = "cat.jpg"

im = Image.open(img_path).convert('L')
w, h = im.size
print(f"התמונה נטענה בהצלחה: רוחב={w}, גובה={h} פיקסלים")
```

## 2. אלגוריתם היפוך תמונה אנכי (Upside Down)

```python
im_inverted = Image.new('L', (w, h), 255)
orig_mat = im.load()
new_mat = im_inverted.load()

for x in range(w):
    for y in range(h):
        new_mat[x, y] = orig_mat[x, h - 1 - y]

print("התמונה נהפכה אנכית בהצלחה!")
# להצגה ב-Jupyter:
im_inverted
```

## 3. אפקט תשליל (Negative)

```python
im_neg = Image.new('L', (w, h), 255)
neg_mat = im_neg.load()

for x in range(w):
    for y in range(h):
        neg_mat[x, y] = 255 - orig_mat[x, y]

print("תשליל הושלם!")
im_neg
```
