## 1. רשימות (Lists) ופעולות עליהן
נבדוק יצירת רשימה, שינוי איבר (mutability) ומיון:

```python
grades = [88, 92, 79, 100, 65]
print("רשימה מקורית:", grades)

# שינוי איבר במקום (Mutable)
grades[0] = 95
print("לאחר שינוי איבר ראשון:", grades)

print("אורך הרשימה:", len(grades))
print("ציון מינימלי:", min(grades))
print("ציון מקסימלי:", max(grades))
print("עותק ממוין:", sorted(grades))
```

## 2. רשימות מקוננות (לוח משחק / מטריצה)

```python
board = [
    ["X", "O", "X"],
    ["O", "X", " "],
    [" ", "O", "O"]
]

print("שורה אמצעית:", board[1])
print("איבר במרכז (שורה 1, עמודה 1):", board[1][1])
```

## 3. משפטי תנאי ופונקציות עצמאיות

```python
def describe_grade(score):
    if score >= 90:
        return "מצוין"
    elif score >= 80:
        return "טוב מאוד"
    elif score >= 60:
        return "עובר"
    else:
        return "נכשל"

print(describe_grade(95))
print(describe_grade(75))
print(describe_grade(50))
```
