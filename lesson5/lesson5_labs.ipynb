## 1. טעינת קובץ נתוני הצעדים (CSV)
נטען את הקובץ ונקרא את שורות הנתונים:

```python
import csv
import os

# איתור מיקום הקובץ (מקומי או ב-Colab)
csv_path = "../data/steps_per_minute.csv"
if not os.path.exists(csv_path):
    csv_path = "data/steps_per_minute.csv"
if not os.path.exists(csv_path):
    # הורדה אוטומטית מ-GitHub אם רץ ב-Colab/Kaggle:
    import urllib.request
    url = "https://raw.githubusercontent.com/alonitzkovitch/python_course_book/main/data/steps_per_minute.csv"
    urllib.request.urlretrieve(url, "steps_per_minute.csv")
    csv_path = "steps_per_minute.csv"

all_data = []
with open(csv_path, mode='r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        all_data.append({"time": row["ActivityMinute"], "steps": int(row["Steps"])})

print(f"נטענו בהצלחה {len(all_data):,} דקות מדידה!")
```

## 2. חישובים סטטיסטיים ומציאת שיא פעילות

```python
total_steps = sum(r["steps"] for r in all_data)
avg_steps = total_steps / len(all_data)

print(f"סך כל הצעדים: {total_steps:,}")
print(f"ממוצע צעדים לדקה: {avg_steps:.2f}")

# מציאת הדקה עם מספר הצעדים המקסימלי
peak = max(all_data, key=lambda r: r["steps"])
print(f"דקת שיא הפעילות: {peak['time']} עם {peak['steps']} צעדים!")
```
