## 1. פעולות חשבון והדפסות ראשונות
הריצו את התא הבא כדי לראות את פקודת `print()` בפעולה:

```python
# הדפסה בסיסית
print("Hello, World!")
print("ברוכים הבאים לקורס פייתון!")

# פעולות חשבון
print("10 + 5 =", 10 + 5)
print("21 // 4 (חילוק שלם) =", 21 // 4)
print("21 % 4 (שארית מודולו) =", 21 % 4)
print("2 ** 8 (חזקה) =", 2 ** 8)
```

## 2. משתנים וטיפוסי נתונים
נבדוק את הטיפוס של ערכים שונים באמצעות `type()`:

```python
age = 20
height = 1.78
name = "דנה"
is_student = True

print(age, type(age))
print(height, type(height))
print(name, type(name))
print(is_student, type(is_student))
```

## 3. פעולות על מחרוזות (שרשור, שכפול ואינדקסים)

```python
s1 = "מדעי "
s2 = "המחשב"
s3 = s1 + s2  # שרשור
print(s3)
print("פייתון! " * 3)  # שכפול
print("אורך המחרוזת:", len(s3))
print("תו ראשון:", s3[0])
print("האם מחשב בתוך s3?", "מחשב" in s3)
```

## 4. תרגיל מעשי עצמאי
כתבו בתא הבא קוד שמחשב את סך הדקות ב-3 ימים, והדפיסו את התוצאה:

```python
days = 3
# השלימו את הקוד כאן:
hours = days * 24
minutes = hours * 60
print(f"ב-{days} ימים יש {minutes:,} דקות.")
```
