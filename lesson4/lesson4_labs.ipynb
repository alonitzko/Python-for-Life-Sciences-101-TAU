## 1. יצירת מילונים ופעולות בסיסיות

```python
phonebook = {
    "דנה": "050-1111111",
    "יוסי": "052-2222222",
    "מיכל": "054-3333333"
}

print("טלפון של דנה:", phonebook["דנה"])
# שימוש בטוח ב-get:
print("טלפון של איתן:", phonebook.get("איתן", "לא נמצא"))

# הוספת איבר חדש
phonebook["איתן"] = "058-4444444"
print("ספר הטלפונים המעודכן:", phonebook)
```

## 2. מילון כטבלת שכיחויות (Frequency Table)
נספור את כמות המופעים של כל מילה בטקסט:

```python
quote = "to be or not to be that is the question"
counts = {}

for word in quote.split():
    counts[word] = counts.get(word, 0) + 1

for w, c in counts.items():
    print(f"{w:10} -> {c} פעמים")
```
