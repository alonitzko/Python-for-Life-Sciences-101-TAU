## תרגיל 1: בדיקת פלינדרום
כתבו פונקציה `is_palindrome(s)` המחזירה `True` אם המחרוזת היא פלינדרום, תוך התעלמות מרווחים ואותיות רישיות.

```python
def is_palindrome(s):
    # מחק את pass וכתוב את הפתרון שלך כאן:
    clean = [ch.lower() for ch in s if ch.isalnum()]
    return clean == clean[::-1]

# בדיקות אוטומטיות:
assert is_palindrome('radar') == True
assert is_palindrome('hello') == False
assert is_palindrome('A man a plan a canal Panama') == True
print('תרגיל 1 עבר בהצלחה!')
```

## תרגיל 2: האיבר השני בגודלו
כתבו פונקציה `second_largest(numbers)` שמחזירה את האיבר השני בגודלו ללא כפילויות, או `None` אם אין כזה.

```python
def second_largest(numbers):
    first, second = None, None
    for n in numbers:
        if first is None or n > first:
            second = first
            first = n
        elif n != first and (second is None or n > second):
            second = n
    return second

# בדיקות אוטומטיות:
assert second_largest([10, 40, 20, 50, 30]) == 40
assert second_largest([5, 5, 5]) == None
assert second_largest([100, 20, 100, 40]) == 40
print('תרגיל 2 עבר בהצלחה!')
```

## תרגיל 3: איחוד מילוני מלאי
כתבו פונקציה `merge_inventories(inv1, inv2)` שמחזירה מילון מאוחד עם סכום הכמויות עבור מוצרים משותפים.

```python
def merge_inventories(inv1, inv2):
    res = inv1.copy()
    for k, v in inv2.items():
        res[k] = res.get(k, 0) + v
    return res

# בדיקות אוטומטיות:
b1 = {'apples': 10, 'bananas': 5}
b2 = {'bananas': 15, 'oranges': 8}
merged = merge_inventories(b1, b2)
assert merged == {'apples': 10, 'bananas': 20, 'oranges': 8}
print('תרגיל 3 עבר בהצלחה!')
```
