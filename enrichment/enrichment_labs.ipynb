## 1. השוואת ביצועים: חיפוש סדרתי מול חיפוש בינארי

```python
import time

data = list(range(1_000_000))
target = 999_999

# חיפוש סדרתי
t0 = time.time()
found_linear = False
for x in data:
    if x == target:
        found_linear = True
        break
t_linear = time.time() - t0

# חיפוש בינארי
t0 = time.time()
left, right = 0, len(data) - 1
found_bin = False
while left <= right:
    mid = (left + right) // 2
    if data[mid] == target:
        found_bin = True
        break
    elif target < data[mid]:
        right = mid - 1
    else:
        left = mid + 1
t_bin = time.time() - t0

print(f"זמן חיפוש סדרתי: {t_linear:.6f} שניות")
print(f"זמן חיפוש בינארי: {t_bin:.6f} שניות")
if t_bin > 0:
    print(f"חיפוש בינארי היה מהיר פי {t_linear / t_bin:.1f}!")
```

## 2. שבירת צופן קיסר בסריקה מלאה (Brute Force)

```python
def caesar_decrypt(ciphertext, offset):
    alphabet = "abcdefghijklmnopqrstuvwxyz"
    plain = ""
    for char in ciphertext:
        if char in alphabet:
            pos = alphabet.find(char)
            plain += alphabet[(pos - offset) % 26]
        else:
            plain += char
    return plain

secret = "wklv lv d vhfuhw phvvdjh"
print("שבירת צופן קיסר:")
for offset in range(1, 26):
    guess = caesar_decrypt(secret, offset)
    print(f"היסט {offset:2}: {guess}")
```
