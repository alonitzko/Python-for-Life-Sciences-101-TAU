## 1. לולאות `for` ופונקציית `range`

```python
# הדפסת מספרים זוגיים מ-0 עד 10
for i in range(0, 11, 2):
    print(i, end=" ")
print()

# חישוב סכום איברים ברשימה בעזרת לולאה
numbers = [10, 20, 30, 40, 50]
total = 0
for num in numbers:
    total += num
print("סכום האיברים:", total)
```

## 2. לולאות מקוננות (לוח הכפל)

```python
for row in range(1, 6):
    line = ""
    for col in range(1, 6):
        line += f"{row * col:4}"
    print(line)
```

## 3. מדידת זמני ריצה ($O(n)$) עם מודול `time`

```python
import time

n = 1_000_000
start_time = time.time()

total = 0
for i in range(n):
    total += i

elapsed = time.time() - start_time
print(f"זמן ריצה לסכימת {n:,} מספרים: {elapsed:.4f} שניות")
```
