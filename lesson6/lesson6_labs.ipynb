## 1. גרף קווי בסיסי עם `matplotlib.pyplot`

```python
import matplotlib.pyplot as plt

hours = [0, 4, 8, 12, 16, 20, 24]
temperatures = [14, 12, 18, 26, 24, 19, 15]

plt.figure(figsize=(7, 4))
plt.plot(hours, temperatures, color='blue', marker='o', linewidth=2)
plt.title("טמפרטורה לאורך היממה")
plt.xlabel("שעה")
plt.ylabel("מעלות צלזיוס")
plt.grid(True)
plt.show()
```

## 2. דיאגרמת עמודות והיסטוגרמה

```python
categories = ["A", "B", "C", "D"]
values = [45, 80, 60, 30]

plt.figure(figsize=(6, 3.5))
plt.bar(categories, values, color='teal')
plt.title("השוואת נתוני קטגוריות")
plt.show()
```
